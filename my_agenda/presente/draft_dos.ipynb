```python
# Contenido de prueba para nuestro archivo .env
contenido_env = """# Variables de entorno locales (NO SUBIR A GITHUB)
DOCKER_HUB_USER="leopinzon_demo"
DOCKER_HUB_PASSWORD="mi_token_super_secreto_123"
"""

# Guardamos el archivo .env en la raíz
with open(".env", "w", encoding="utf-8") as f:
    f.write(contenido_env)

print("✅ Archivo .env creado con éxito en la raíz de tu proyecto.")
```

```python
import os
from dotenv import load_dotenv

# 1. Cargamos las variables guardadas en el archivo .env a la memoria
load_dotenv()

# 2. Leemos las variables usando os.getenv()
usuario = os.getenv("DOCKER_HUB_USER")
password = os.getenv("DOCKER_HUB_PASSWORD")

# 3. Comprobamos que Python las reconoce
print(f"👤 Usuario cargado: {usuario}")
print(f"🔑 Longitud de la clave cargada: {len(password)} caracteres")
print(f"🛡️ Máscara de seguridad: {'*' * len(password)}")
```

```python
import os
from dotenv import load_dotenv

# Aseguramos que las variables del .env estén en el entorno del sistema (os.environ)
load_dotenv()

# Asignamos las variables de Python al entorno de la sesión para que la terminal las reconozca
os.environ["DOCKER_USER"] = os.getenv("DOCKER_HUB_USER")
os.environ["DOCKER_PASS"] = os.getenv("DOCKER_HUB_PASSWORD")

print("🧪 Probando simulación de login con variables de entorno...")

# Simulamos la autenticación leyendo la variable desde la terminal con 'echo'
# (Esta misma estructura se usa con 'docker login -u "$DOCKER_USER" --password-stdin')
!echo "Autenticando usuario $DOCKER_USER sin mostrar el token en pantalla..."
```

```python
import subprocess

def ejecutar_comando_git(cmd):
    resultado = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    return resultado.stdout.strip()

print("🔍 1. Verificando que Git IGNORE el archivo .env...")
status = ejecutar_comando_git("git status")

if ".env" in status:
    print("⚠️  ATENCIÓN: El archivo .env aparece en git status. Hay que revisar el .gitignore.")
else:
    print("✅ ¡Éxito! El archivo .env está oculto correctamente y Git lo ignoró.")

print("\n🚀 2. Guardando cambios de esta práctica en GitHub...")
ejecutar_comando_git("git add .")
ejecutar_comando_git('git commit -m "práctica: implementación exitosa de variables de entorno con .env"')
salida_push = ejecutar_comando_git("git push origin main")

print("📡 Resultado del push a GitHub:")
print(salida_push if salida_push else "✅ ¡Push completado sin alertas ni bloqueos de seguridad!")
```

```python
import os
import shutil

trash_dir = os.path.expanduser("~/.Trash")
target_dir = "/Users/admin/Desktop/Shafer_Python_Classes/Cloud_engineering_practice/my_agenda/presente"

os.makedirs(target_dir, exist_ok=True)

print("🔎 Buscando archivos en la Papelera...\n")

found_files = []

# Recorrer minuciosamente toda la papelera
for root, dirs, files in os.walk(trash_dir):
    for file in files:
        # Filtramos por cuadernos de Jupyter o cualquier archivo con palabras clave de tu agenda
        if file.endswith(".ipynb") or "draft" in file.lower() or "dreft" in file.lower() or "presente" in file.lower():
            full_path = os.path.join(root, file)
            found_files.append((full_path, file))

if not found_files:
    print("⚠️ No se encontraron archivos .ipynb directos. Verificando carpetas completas en Trash...")
    for item in os.listdir(trash_dir):
        if "Cloud_engineering" in item or "my_agenda" in item or "Shafer" in item:
            print(f"📁 Carpeta de respaldo encontrada en Trash: {item}")
else:
    recovered = 0
    for src_path, file_name in found_files:
        # Limpiar el nombre para quitar la marca '-checkpoint' si existe
        clean_name = file_name.replace("-checkpoint", "")
        if not clean_name.endswith(".ipynb"):
            clean_name += ".ipynb"
            
        dest_path = os.path.join(target_dir, clean_name)
        
        try:
            shutil.copy2(src_path, dest_path)
            print(f"✅ RECUPERADO: {clean_name}")
            print(f"   Ubicación original en Trash: {src_path}\n")
            recovered += 1
        except Exception as e:
            print(f"❌ Error al copiar {file_name}: {e}")

    print(f"🎉 Total de archivos recuperados: {recovered}")
```

```python
import os
import re

# Palabras clave y patrones de secretos comunes
patrones_secretos = [
    (r'ghp_[a-zA-Z0-9]{36}', 'Token de GitHub'),
    (r'dckr_pat_[a-zA-Z0-9_\-]{30,}', 'Token de Docker Hub'),
    (r'AKIA[0-9A-Z]{16}', 'Clave de AWS Access Key'),
    (r'(?i)(password|passwd|secret|api_key)\s*=\s*["\'][^"\']{8,}["\']', 'Posible contraseña o API Key explícita')
]

carpeta_proyecto = "/Users/admin/Desktop/Shafer_Python_Classes/Cloud_engineering_practice"

print("🛡️ Iniciando escaneo de seguridad en tus cuadernos `.ipynb`...\n")
alertas_encontradas = 0

for root, dirs, files in os.walk(carpeta_proyecto):
    # Ignorar carpetas de checkpoint y entornos virtuales
    if ".ipynb_checkpoints" in root or "venv" in root or ".git" in root:
        continue
        
    for file in files:
        if file.endswith(".ipynb"):
            ruta_completa = os.path.join(root, file)
            try:
                with open(ruta_completa, "r", encoding="utf-8") as f:
                    contenido = f.read()
                    
                    for patron, tipo in patrones_secretos:
                        coincidencias = re.findall(patron, contenido)
                        if coincidencias:
                            print(f"⚠️  ALERTA DE SEGURIDAD en: {os.path.relpath(ruta_completa, carpeta_proyecto)}")
                            print(f"    - Tipo detectado: {tipo}")
                            alertas_encontradas += 1
            except Exception as e:
                pass

if alertas_encontradas == 0:
    print("✅ ¡Escaneo completado con éxito! No se detectaron tokens ni contraseñas expuestas en tus cuadernos.")
else:
    print(f"\n⚠️  Se encontraron {alertas_encontradas} advertencias. Revisa los cuadernos señalados antes de hacer git push.")
    
```

```python
import os
import re

patrones_secretos = [
    (r'ghp_[a-zA-Z0-9]{36}', 'Token de GitHub'),
    (r'dckr_pat_[a-zA-Z0-9_\-]{30,}', 'Token de Docker Hub'),
    (r'AKIA[0-9A-Z]{16}', 'Clave de AWS Access Key')
]

carpeta_proyecto = "/Users/admin/Desktop/Shafer_Python_Classes/Cloud_engineering_practice"

print("🛡️ Verificando seguridad final...\n")
alertas_encontradas = 0

for root, dirs, files in os.walk(carpeta_proyecto):
    if ".ipynb_checkpoints" in root or "venv" in root or ".git" in root:
        continue
        
    for file in files:
        if file.endswith(".ipynb"):
            ruta_completa = os.path.join(root, file)
            try:
                with open(ruta_completa, "r", encoding="utf-8") as f:
                    contenido = f.read()
                    
                    for patron, tipo in patrones_secretos:
                        coincidencias = re.findall(patron, contenido)
                        if coincidencias:
                            print(f"⚠️ ALERTA en: {os.path.relpath(ruta_completa, carpeta_proyecto)} ({tipo})")
                            alertas_encontradas += 1
            except Exception:
                pass

if alertas_encontradas == 0:
    print("✅ ¡Escaneo limpio! No se encontraron credenciales reales expuestas. Tu repositorio es 100% seguro para GitHub.")
```

```python
import subprocess

def ejecutar_comando_git(cmd):
    print(f"🚀 Ejecutando: {cmd}")
    resultado = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    if resultado.stdout:
        print(resultado.stdout)
    if resultado.stderr:
        print(resultado.stderr)
    return resultado.returncode

# 1. Agregar todos los cuadernos y cambios respetando el .gitignore
ejecutar_comando_git("git add .")

# 2. Crear el commit con un mensaje descriptivo
mensaje = "Actualización de seguridad, recuperación de cuadernos de trabajo y notas de Cloud Engineering"
ejecutar_comando_git(f'git commit -m "{mensaje}"')

# 3. Enviar los cambios a GitHub
print("\n📡 Subiendo cambios a GitHub...")
codigo_salida = ejecutar_comando_git("git push origin main")

if codigo_salida == 0:
    print("\n🎉 ¡ÉXITO TOTAL! Todo tu material resguardado y respaldado en tu repositorio de GitHub.")
else:
    print("\n⚠️ Nota: Si tu rama principal se llama 'master' en lugar de 'main', probando 'git push origin master'...")
    ejecutar_comando_git("git push origin master")
```

```python
import subprocess

def guardar_proyecto_completo(mensaje_commit="feat: avance diario colectivo en el proyecto"):
    print("📦 1. Escaneando y agregando todos los cuadernos y archivos modificados...")
    subprocess.run("git add .", shell=True, check=True)
    
    print("📝 2. Creando la instantánea colectiva (commit)...")
    # Capturamos el resultado por si no hay cambios nuevos que guardar
    resultado_commit = subprocess.run(f'git commit -m "{mensaje_commit}"', shell=True, capture_output=True, text=True)
    
    if "nothing to commit" in resultado_commit.stdout:
        print("ℹ️ No hay cambios nuevos registrados para guardar.")
        return
    else:
        print(resultado_commit.stdout.strip())
        
    print("\n📡 3. Subiendo todos los cambios colectivos a GitHub...")
    resultado_push = subprocess.run("git push origin main", shell=True, capture_output=True, text=True)
    
    if resultado_push.returncode == 0:
        print("\n🎉 ¡ÉXITO TOTAL! Todos los notebooks y avances del día quedaron resguardados en GitHub.")
    else:
        print(f"\n⚠️ Error o advertencia al hacer push:\n{resultado_push.stderr}")

# Ejecutas la función al terminar tu jornada
guardar_proyecto_completo("feat: avance colectivo del día en cuadernos y notas de cloud engineering")
```

```python
import subprocess

print("🔍 Escaneando la base de datos de Git por archivos borrados que contengan 'objetos'...\n")

# Comando para buscar en todos los commits y blobs de Git la palabra 'objetos'
comando_git = 'git grep -i "objetos" $(git rev-list --all)'
resultado = subprocess.getoutput(comando_git)

rutas_encontradas = set()

for linea in resultado.split('\n'):
    if linea and ":" in linea:
        # Extraemos el identificador/ruta del archivo en Git
        partes = linea.split(":", 2)
        ref_y_ruta = partes[0] + ":" + partes[1] if len(partes) > 2 else partes[0]
        
        # Filtramos para NO mostrar carpetas 'recuperado' ni checkpoints
        if "recuperado" not in ref_y_ruta.lower() and ".ipynb_checkpoints" not in ref_y_ruta:
            rutas_encontradas.add(ref_y_ruta)

print(f"📌 Se encontraron {len(rutas_encontradas)} referencias únicas en el historial de Git:\n")
for r in sorted(rutas_encontradas):
    print(r)
```

```python
import json

# Ruta al archivo de 63KB
ruta_draft_dos = "/Users/admin/Desktop/Shafer_Python_Classes/Cloud_engineering_practice/my_agenda/presente/draft_dos.ipynb"

try:
    with open(ruta_draft_dos, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    celdas = data.get('cells', [])
    print(f"📊 Total de celdas en el archivo: {len(celdas)}\n")
    print("--- PRIMERAS 5 CELDAS DEL ARCHIVO DE 63KB ---")
    
    # Mostramos los primeros textos/códigos guardados dentro del cuaderno
    for i, celda in enumerate(celdas[:5]):
        tipo = celda.get('cell_type', 'desconocido')
        fuente = "".join(celda.get('source', []))
        print(f"\n[Celda {i+1} - {tipo.upper()}]:")
        print(fuente[:300] + "..." if len(fuente) > 300 else fuente)

except Exception as e:
    print(f"Error al leer el cuaderno: {e}")
```
